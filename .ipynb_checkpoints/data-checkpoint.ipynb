```python
# Process images
from PIL import Image 
import numpy as np
import torchvision.transforms as T

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

transform = T.Compose([
        T.Resize((128, 128)), #resize to 64x64
        T.CenterCrop(128), 
        T.RandomHorizontalFlip(p=0.5), 
        T.ToTensor(),
        T.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))#normalize

        # T.RandomHorizontalFlip(p=0.5), #data augmentation
        # T.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.05), #data augmentation
        # T.RandomRotation(degrees=7), #data augmentation
        # T.RandomPerspective(distortion_scale=0.05, p=0.5), #data augmentation

        # T.ToTensor(),
        # T.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))#normalize
])

for key, value in data_dict.items():
    img = Image.open(key).convert('RGB')  
    img_tensor = transform(img).to(device)   
    data_dict[key]['image_tensor'] = img_tensor
    
    img = torch.tensor(np.array(img)).permute(2, 0, 1).float() / 255.0 #i removed unsqueeze because it was givinig mismatched tensor size errors in the training loop
    img = img.to(device)
    # print(f"Image Path: {key}, Image Shape: {img.shape}, type: {type(img)}")

    zeros_class = torch.zeros(4, dtype=torch.float32).to(device)
    zeros_country = torch.zeros(48, dtype=torch.float32).to(device)
    zeros_major = torch.zeros(42, dtype=torch.float32).to(device)

    class_label_one_hot = zeros_class.clone()
    class_label_one_hot[value['class_label']] = 1.0
    country_label_one_hot = zeros_country.clone()
    country_label_one_hot[value['country_labels']] = 1.0   
    major_label_one_hot = zeros_major.clone() #this shoudl have been outside loop
    for major_label in value['major_labels']:
        major_label_one_hot[major_label] = 1.0

    # print(f"Class Label One-Hot: {class_label_one_hot}, Country Label One-Hot: {country_label_one_hot}, Major Label One-Hot: {major_label_one_hot}")
    

    # store in dict 
    data_dict[key]['image_tensor'] = img_tensor
    data_dict[key]['class_label_one_hot'] = class_label_one_hot
    data_dict[key]['country_label_one_hot'] = country_label_one_hot
    data_dict[key]['major_label_one_hot'] = major_label_one_hot
    
    
```

```python
class ClassMates(Dataset):
    def __init__(self, data_dict):
        self.data_dict = data_dict
        self.image_paths = list(data_dict.keys())
        
    def __len__(self):
        return len(self.image_paths)
    
    def __getitem__(self, idx):
        image_path = self.image_paths[idx]
        data = self.data_dict[image_path]
        return data['image_tensor'], data['class_label_one_hot'], data['country_label_one_hot'], data['major_label_one_hot']
```

```python
dataset = ClassMates(data_dict)
dataloader = DataLoader(dataset, batch_size=64, shuffle=True, drop_last=True)

for batch in dataloader:
    images, class_labels, country_labels, major_labels = batch
    print(f"Batch Image Shape: {images.shape}")
    print(f"Batch Class Labels Shape: {class_labels.shape}")
    print(f"Batch Country Labels Shape: {country_labels.shape}")
    print(f"Batch Major Labels Shape: {major_labels.shape}")

    break
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import os
from torchvision.utils import make_grid
import matplotlib.pyplot as plt
import torch.nn.utils.spectral_norm as spectral_norm


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

latent_dim = 100
condition_dim = 94 # 4 (Class) + 48 (Country) + 42 (Major)
channels = 3
img_size = 128 #resized down to make it easier but native is 600 X 800

class DCGAN_gen(nn.Module):
    def __init__(self):
        super().__init__()
        self.model = nn.Sequential(
            # Input: (batch, 194, 1, 1)
            nn.ConvTranspose2d(latent_dim + condition_dim, 256, kernel_size=4, stride=1, padding=0, bias=False),
            nn.BatchNorm2d(256),
            nn.ReLU(True),
            # Size: (batch, 256, 4, 4)

            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv2d(256, 128, kernel_size=3, stride=1, padding=1, bias=False),
            nn.BatchNorm2d(128),  
            nn.LeakyReLU(0.2, inplace=True),
            # Size: (batch, 128, 8, 8)

            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv2d(128, 64, kernel_size=3, stride=1, padding=1, bias=False),
            nn.BatchNorm2d(64),  # FIX 1: was BatchNorm2d
            nn.LeakyReLU(0.2, inplace=True),
            # Size: (batch, 64, 16, 16)


            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv2d(64, 32, kernel_size=3, stride=1, padding=1, bias=False),
            nn.BatchNorm2d(32),  # FIX 1: was BatchNorm2d
            nn.LeakyReLU(0.2, inplace=True),
            # Size: (batch, 32, 32, 32)

            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv2d(32, 16, kernel_size=3, stride=1, padding=1, bias=False),
            nn.BatchNorm2d(16),  # FIX 1: was BatchNorm2d
            nn.LeakyReLU(0.2, inplace=True),
            # Size: (batch, 16, 64, 64)

            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv2d(16, 3, kernel_size=3, stride=1, padding=1, bias=False),
            # Size: (batch, 3, 128, 128)
        
            nn.Tanh()
            # Final Size: (batch, 3, 128, 128)
        )

    def forward(self, noise, combined_labels):
        c = combined_labels.view(combined_labels.size(0), combined_labels.size(1), 1, 1)
        x = torch.cat([noise, c], dim=1)
        return self.model(x)


# --- DISCRIMINATOR ---
class DCGAN_disc(nn.Module):
    def __init__(self):
        super().__init__()
        
        # Project the 94-dimensional condition down to 16 dimensions
        self.condition_project = nn.Sequential(
            spectral_norm(nn.Linear(condition_dim, 16)),
            nn.LeakyReLU(0.2, inplace=True)
        )

        self.conv1 = nn.Sequential(
            spectral_norm(nn.Conv2d(channels, 64, kernel_size=4, stride=2, padding=1, bias=False)),
            nn.LeakyReLU(0.2, inplace=True)
        )
        
        self.conv2 = nn.Sequential(
            spectral_norm(nn.Conv2d(64 + 16, 128, kernel_size=4, stride=2, padding=1, bias=False)),
            nn.LeakyReLU(0.2, inplace=True),
            
            spectral_norm(nn.Conv2d(128, 256, kernel_size=4, stride=2, padding=1, bias=False)),
            nn.LeakyReLU(0.2, inplace=True),
            
            spectral_norm(nn.Conv2d(256, 512, kernel_size=4, stride=2, padding=1, bias=False)),
            nn.LeakyReLU(0.2, inplace=True),

            spectral_norm(nn.Conv2d(512, 512, kernel_size=4, stride=2, padding=1, bias=False)),
            nn.LeakyReLU(0.2, inplace=True),
            
            spectral_norm(nn.Conv2d(512, 1, kernel_size=4, stride=1, padding=0, bias=False))
        )

    def forward(self, img, combined_labels):
        # 1. Pass image through first conv layer
        x = self.conv1(img)
        
        # 2. Project the condition labels from 94 down to 16
        c_proj = self.condition_project(combined_labels) 
        
        # 3. Reshape and expand the condition to match the spatial dimensions of x (64x64)
        c = c_proj.view(c_proj.size(0), c_proj.size(1), 1, 1)
        c = c.expand(-1, -1, x.size(2), x.size(3))
        
        # 4. Concatenate along the channel dimension (dim=1) -> 64 + 16 = 80 channels
        x = torch.cat([x, c], dim=1)
        
        # 5. Pass through the rest of the discriminator
        validity = self.conv2(x)
        
        return validity.view(-1, 1) # Flatten to (batch, 1)
```

```python
# from models import DCGAN_gen, DCGAN_disc
import torch
import torch.nn as nn
import torch.optim as optim

# Initialize models
generator_DCGAN = DCGAN_gen().to(device)
discriminator_DCGAN = DCGAN_disc().to(device)

latent_dim = 100
condition_dim = 94 # 4 (Class) + 48 (Country) + 42 (Major)
channels = 3
img_size = 128
```

```python
# Initialize Models
generator_DCGAN = DCGAN_gen().to(device)
discriminator_DCGAN = DCGAN_disc().to(device)


def weights_init(m):
    classname = m.__class__.__name__
    if classname.find('Conv') != -1:
        nn.init.normal_(m.weight.data, 0.0, 0.02)
    elif classname.find('BatchNorm') != -1:
        nn.init.normal_(m.weight.data, 1.0, 0.02)
        nn.init.constant_(m.bias.data, 0)
    elif classname.find('Linear') != -1: # Added this to catch the condition_project layer
        nn.init.normal_(m.weight.data, 0.0, 0.02)
        nn.init.constant_(m.bias.data, 0)

# Apply the initialization right after defining the models
generator_DCGAN.apply(weights_init)
discriminator_DCGAN.apply(weights_init)


# Try lowering the learning rates and keeping them equal
d_optimizer = optim.Adam(discriminator_DCGAN.parameters(), lr=0.0002, betas=(0.5, 0.999))
g_optimizer = optim.Adam(generator_DCGAN.parameters(), lr=0.0002, betas=(0.5, 0.999))

loss_fn = nn.BCEWithLogitsLoss() 

def noise_2d(batch_size):
    return torch.randn(batch_size, latent_dim, 1, 1, device=device)
```

```python
import matplotlib.pyplot as plt
from torchvision.utils import make_grid

def show_generated_images(generator, num_images=16):
    generator.eval() # Put generator in evaluation mode
    
    with torch.no_grad():
        z = noise_2d(num_images)
        zeros_class = torch.zeros(num_images, 4, device=device)
        zeros_country = torch.zeros(num_images, 48, device=device)
        zeros_major = torch.zeros(num_images, 42, device=device)
        
        for i in range(num_images):
            rand_class = torch.randint(0, 4, (1,)).item()
            rand_country = torch.randint(0, 48, (1,)).item()
            rand_major = torch.randint(0, 42, (1,)).item()
            
            zeros_class[i, rand_class] = 1.0
            zeros_country[i, rand_country] = 1.0
            zeros_major[i, rand_major] = 1.0
            
        fake_conditions = torch.cat([zeros_class, zeros_country, zeros_major], dim=1)
        
        fake_images = generator(z, fake_conditions)
        
        fake_images = (fake_images + 1) / 2.0
        grid = make_grid(fake_images.cpu(), nrow=4, padding=2, normalize=False)
        
        plt.figure(figsize=(8, 8))
        plt.axis("off")
        plt.title("Generated Faces")
        plt.imshow(grid.permute(1, 2, 0)) 
        plt.show()
        
    generator.train() # Put generator back into training mode
```

## Transfer Learning

```python
import torchvision.models as models

# 1. Load a pre-trained VGG-16 model
# We use the 'features' portion to extract high-level textures rather than classification labels
vgg = models.vgg16(pretrained=True).features.to(device)

# 2. We only need the first few layers (which detect edges, textures, and facial structures)
# Layer 16 corresponds to the relu3_3 layer, standard for perceptual loss
vgg_expert = nn.Sequential(*list(vgg.children())[:16]).eval()

# 3. Freeze the VGG weights so they don't update during GAN training
for param in vgg_expert.parameters():
    param.requires_grad = False

# Perceptual loss is typically calculated using Mean Squared Error
perceptual_loss_fn = nn.MSELoss()

# VGG expects images normalized with ImageNet mean/std, so we need a quick normalizer
def normalize_for_vgg(x):
    # GAN output is [-1, 1]. Shift to [0, 1]
    x = (x + 1.0) / 2.0 
    
    # ImageNet normalization constants
    mean = torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1).to(x.device)
    std = torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1).to(x.device)
    return (x - mean) / std
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision.transforms as T
import torchvision.datasets as datasets
from torch.utils.data import DataLoader
from torchvision.utils import make_grid
import matplotlib.pyplot as plt
import torch.nn.utils.spectral_norm as spectral_norm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device = 'cuda'
# ==========================================
#             1. HYPERPARAMETERS
# ==========================================
latent_dim = 100
condition_dim = 94 # 4 (Class) + 48 (Country) + 42 (Major)
channels = 3
img_size = 128
batch_size = 3500 # Increased for faster training on high-end GPUs
num_epochs = 50

# ==========================================
#             2. MODEL DEFINITIONS
# ==========================================
class DCGAN_gen(nn.Module):
    def __init__(self):
        super().__init__()
        self.model = nn.Sequential(
            nn.ConvTranspose2d(latent_dim + condition_dim, 256, kernel_size=4, stride=1, padding=0, bias=False),
            nn.BatchNorm2d(256),
            nn.ReLU(True),
            
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv2d(256, 128, kernel_size=3, stride=1, padding=1, bias=False),
            nn.BatchNorm2d(128),  
            nn.LeakyReLU(0.2, inplace=True),
            
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv2d(128, 64, kernel_size=3, stride=1, padding=1, bias=False),
            nn.BatchNorm2d(64),  
            nn.LeakyReLU(0.2, inplace=True),
            
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv2d(64, 32, kernel_size=3, stride=1, padding=1, bias=False),
            nn.BatchNorm2d(32),  
            nn.LeakyReLU(0.2, inplace=True),
            
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv2d(32, 16, kernel_size=3, stride=1, padding=1, bias=False),
            nn.BatchNorm2d(16),  
            nn.LeakyReLU(0.2, inplace=True),
            
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv2d(16, 3, kernel_size=3, stride=1, padding=1, bias=False),
            nn.Tanh()
        )

    def forward(self, noise, combined_labels):
        c = combined_labels.view(combined_labels.size(0), combined_labels.size(1), 1, 1)
        x = torch.cat([noise, c], dim=1)
        return self.model(x)


class DCGAN_disc(nn.Module):
    def __init__(self):
        super().__init__()
        
        self.condition_project = nn.Sequential(
            spectral_norm(nn.Linear(condition_dim, 16)),
            nn.LeakyReLU(0.2, inplace=True)
        )

        self.conv1 = nn.Sequential(
            spectral_norm(nn.Conv2d(channels, 64, kernel_size=4, stride=2, padding=1, bias=False)),
            nn.LeakyReLU(0.2, inplace=True)
        )
        
        self.conv2 = nn.Sequential(
            spectral_norm(nn.Conv2d(64 + 16, 128, kernel_size=4, stride=2, padding=1, bias=False)),
            nn.LeakyReLU(0.2, inplace=True),
            
            spectral_norm(nn.Conv2d(128, 256, kernel_size=4, stride=2, padding=1, bias=False)),
            nn.LeakyReLU(0.2, inplace=True),
            
            spectral_norm(nn.Conv2d(256, 512, kernel_size=4, stride=2, padding=1, bias=False)),
            nn.LeakyReLU(0.2, inplace=True),

            spectral_norm(nn.Conv2d(512, 512, kernel_size=4, stride=2, padding=1, bias=False)),
            nn.LeakyReLU(0.2, inplace=True),
            
            spectral_norm(nn.Conv2d(512, 1, kernel_size=4, stride=1, padding=0, bias=False))
        )

    def forward(self, img, combined_labels):
        x = self.conv1(img)
        c_proj = self.condition_project(combined_labels) 
        c = c_proj.view(c_proj.size(0), c_proj.size(1), 1, 1)
        c = c.expand(-1, -1, x.size(2), x.size(3))
        
        x = torch.cat([x, c], dim=1)
        validity = self.conv2(x)
        return validity.view(-1, 1)


# ==========================================
#          3. DATASET & DATALOADER
# ==========================================
transform = T.Compose([
    T.Resize((128, 128)),
    T.CenterCrop(128),
    T.RandomHorizontalFlip(p=0.5),
    T.ToTensor(),
    T.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

# NOTE: If PyTorch's auto-download fails due to Google Drive quota limits, 
# download the "img_align_celeba" folder manually and use datasets.ImageFolder instead.
celeba_dataset = datasets.CelebA(root='./', split='train', target_type='attr', download=True, transform=transform)
dataloader = DataLoader(celeba_dataset, batch_size=batch_size, shuffle=True, drop_last=True, num_workers=4)

# ==========================================
#             4. INITIALIZATION
# ==========================================
# generator_DCGAN = DCGAN_gen().to(device)
# discriminator_DCGAN = DCGAN_disc().to(device)

def weights_init(m):
    classname = m.__class__.__name__
    if classname.find('Conv') != -1:
        nn.init.normal_(m.weight.data, 0.0, 0.02)
    elif classname.find('BatchNorm') != -1:
        nn.init.normal_(m.weight.data, 1.0, 0.02)
        nn.init.constant_(m.bias.data, 0)
    elif classname.find('Linear') != -1: 
        nn.init.normal_(m.weight.data, 0.0, 0.02)
        nn.init.constant_(m.bias.data, 0)

# generator_DCGAN.apply(weights_init)
# discriminator_DCGAN.apply(weights_init)

generator_DCGAN = DCGAN_gen().to(device)
discriminator_DCGAN = DCGAN_disc().to(device)


# 2. Load the weight dictionaries into those objects
# This maps the weights from the file into the layers of your model
generator_DCGAN.load_state_dict(torch.load('celeba_gen_epoch_15.pth'))
discriminator_DCGAN.load_state_dict(torch.load('celeba_disc_epoch_15.pth'))

# 3. NOW initialize the optimizers
# Because the models are now objects again, they have .parameters()
d_optimizer = optim.Adam(discriminator_DCGAN.parameters(), lr=0.0002, betas=(0.5, 0.999))
g_optimizer = optim.Adam(generator_DCGAN.parameters(), lr=0.0002, betas=(0.5, 0.999))



def noise_2d(b_size):
    return torch.randn(b_size, latent_dim, 1, 1, device=device)

def show_generated_images(generator, num_images=16):
    generator.eval()
    with torch.no_grad():
        z = noise_2d(num_images)
        
        # Generate random dummy conditions for the preview grid
        dummy_class = torch.zeros(num_images, 4, device=device)
        dummy_country = torch.zeros(num_images, 48, device=device)
        dummy_major = torch.zeros(num_images, 42, device=device)
        
        dummy_class.scatter_(1, torch.randint(0, 4, (num_images, 1), device=device), 1.0)
        dummy_country.scatter_(1, torch.randint(0, 48, (num_images, 1), device=device), 1.0)
        dummy_major.scatter_(1, torch.randint(0, 42, (num_images, 1), device=device), 1.0)
        
        fake_conditions = torch.cat([dummy_class, dummy_country, dummy_major], dim=1)
        
        fake_images = generator(z, fake_conditions)
        fake_images = (fake_images + 1) / 2.0
        grid = make_grid(fake_images.cpu(), nrow=4, padding=2, normalize=False)
        
        plt.figure(figsize=(8, 8))
        plt.axis("off")
        plt.title("CelebA Pre-training Faces")
        plt.imshow(grid.permute(1, 2, 0)) 
        plt.show()
    generator.train()

# ==========================================
#               5. TRAINING LOOP
# ==========================================
start = 15
end = 50 
for epoch in range(15, 50):
    print(f"epoch {epoch + 1}")
    # CelebA returns (image, attributes). We ignore the CelebA attributes.
    for i, (images, _) in enumerate(dataloader):
        n_images = len(images)
        real_data = images.to(device)
        
        # --- GENERATE DUMMY CONDITIONS ON THE FLY ---
        dummy_class = torch.zeros(n_images, 4, device=device)
        dummy_country = torch.zeros(n_images, 48, device=device)
        dummy_major = torch.zeros(n_images, 42, device=device)
        
        dummy_class.scatter_(1, torch.randint(0, 4, (n_images, 1), device=device), 1.0)
        dummy_country.scatter_(1, torch.randint(0, 48, (n_images, 1), device=device), 1.0)
        dummy_major.scatter_(1, torch.randint(0, 42, (n_images, 1), device=device), 1.0)
        
        real_conditions = torch.cat([dummy_class, dummy_country, dummy_major], dim=1)

        # ------------------------------------------
        #            TRAIN DISCRIMINATOR
        # ------------------------------------------
        discriminator_DCGAN.zero_grad()
        
        pred_real = discriminator_DCGAN(real_data, real_conditions)
        loss_real = loss_fn(pred_real, torch.full((n_images, 1), 0.9, device=device)) 
        
        z = noise_2d(n_images)
        fake_data = generator_DCGAN(z, real_conditions)
        
        pred_fake = discriminator_DCGAN(fake_data.detach(), real_conditions)
        loss_fake = loss_fn(pred_fake, torch.full((n_images, 1), 0.1, device=device)) 
        
        d_loss = loss_real + loss_fake
        d_loss.backward()
        d_optimizer.step()

        # ------------------------------------------
        #              TRAIN GENERATOR
        # ------------------------------------------
        g_optimizer.zero_grad()
        
        pred_fake_gen = discriminator_DCGAN(fake_data, real_conditions)
        g_loss = loss_fn(pred_fake_gen, torch.full((n_images, 1), 0.9, device=device))
        
        g_loss.backward()
        g_optimizer.step()
        
        # Optional: Print batch progress for large datasets
        if i % 200 == 0:
            print(f"Epoch {epoch+1} | Batch {i}/{len(dataloader)} | D Loss: {d_loss.item():.4f} | G Loss: {g_loss.item():.4f}")

    # ------------------------------------------
    #              LOGGING & SAVING
    # ------------------------------------------
    if (epoch + 1) % 5 == 0:
        show_generated_images(generator_DCGAN, num_images=16)
        torch.save(generator_DCGAN.state_dict(), f"celeba_gen_epoch_{epoch+1}.pth")
        torch.save(discriminator_DCGAN.state_dict(), f"celeba_disc_epoch_{epoch+1}.pth")

# Final Save
torch.save(generator_DCGAN.state_dict(), "celeba_generator_final.pth")
torch.save(discriminator_DCGAN.state_dict(), "celeba_discriminator_final.pth")
print("CelebA Pre-training Complete!")
```

```python
# ==========================================
#        1. LOAD PRE-TRAINED WEIGHTS
# ==========================================
generator_DCGAN = DCGAN_gen().to(device)
discriminator_DCGAN = DCGAN_disc().to(device)

# Load the weights you saved at the end of Phase 1
generator_DCGAN.load_state_dict(torch.load("celeba_generator_final.pth"))
discriminator_DCGAN.load_state_dict(torch.load("celeba_discriminator_final.pth"))

print("Successfully loaded CelebA pre-trained weights!")

# ==========================================
#     2. FINE-TUNING HYPERPARAMETERS
# ==========================================
# CRITICAL: Drop the learning rate significantly so we don't blow up the learned faces
fine_tune_lr = 0.00005 

d_optimizer = optim.Adam(discriminator_DCGAN.parameters(), lr=fine_tune_lr, betas=(0.5, 0.999))
g_optimizer = optim.Adam(generator_DCGAN.parameters(), lr=fine_tune_lr, betas=(0.5, 0.999))
loss_fn = nn.BCEWithLogitsLoss()

# DiffAugment helper function (re-added for the 400-image dataset)
def diff_augment(x):
    alpha = 1.0 + (torch.rand(x.size(0), 1, 1, 1, device=x.device) - 0.5) * 0.4 # Contrast
    beta = (torch.rand(x.size(0), 1, 1, 1, device=x.device) - 0.5) * 0.4 # Brightness
    return torch.clamp(x * alpha + beta, -1.0, 1.0)

def noise_2d(batch_size):
    return torch.randn(batch_size, latent_dim, 1, 1, device=device)

# ==========================================
#             3. FINE-TUNING LOOP
# ==========================================
# Fine-tuning usually requires fewer epochs. Start with 200 and observe.
num_epochs_finetune = 200 

for epoch in range(num_epochs_finetune):
    # Using your original dataloader with the real Class, Country, and Major labels
    for i, (images, class_labels, country_labels, major_labels) in enumerate(dataloader):
        n_images = len(images)
        
        real_data = images.to(device)
        real_conditions = torch.cat([class_labels, country_labels, major_labels], dim=1).to(device)
        
        # ------------------------------------------
        #            TRAIN DISCRIMINATOR
        # ------------------------------------------
        discriminator_DCGAN.zero_grad()
        
        # 1. Real Data Pass (with DiffAugment)
        aug_real_data = diff_augment(real_data)
        pred_real = discriminator_DCGAN(aug_real_data, real_conditions)
        loss_real = loss_fn(pred_real, torch.full((n_images, 1), 0.9, device=device)) 
        
        # 2. Fake Data Pass (with DiffAugment)
        z = noise_2d(n_images)
        fake_data = generator_DCGAN(z, real_conditions)
        
        aug_fake_data = diff_augment(fake_data.detach())
        pred_fake = discriminator_DCGAN(aug_fake_data, real_conditions)
        loss_fake = loss_fn(pred_fake, torch.full((n_images, 1), 0.1, device=device)) 
        
        d_loss = loss_real + loss_fake
        d_loss.backward()
        d_optimizer.step()

        # ------------------------------------------
        #              TRAIN GENERATOR
        # ------------------------------------------
        g_optimizer.zero_grad()
        
        # Apply DiffAugment so gradients match the Discriminator's evaluation
        aug_fake_data_gen = diff_augment(fake_data)
        pred_fake_gen = discriminator_DCGAN(aug_fake_data_gen, real_conditions)
        
        g_loss = loss_fn(pred_fake_gen, torch.full((n_images, 1), 0.9, device=device))
        
        g_loss.backward()
        g_optimizer.step()

    # ------------------------------------------
    #              LOGGING & SAVING
    # ------------------------------------------
    print(f"Fine-Tuning Epoch {epoch+1}/{num_epochs_finetune} | D Loss: {d_loss.item():.4f} | G Loss: {g_loss.item():.4f}")
    
    # Show images more frequently since we are fine-tuning
    if (epoch + 1) % 5 == 0:
        show_generated_images(generator_DCGAN, num_images=16)

    if (epoch + 1) % 50 == 0:
        torch.save(generator_DCGAN.state_dict(), f"custom_finetuned_gen_epoch_{epoch+1}.pth")
        torch.save(discriminator_DCGAN.state_dict(), f"custom_finetuned_disc_epoch_{epoch+1}.pth")

# Final save
torch.save(generator_DCGAN.state_dict(), "custom_finetuned_generator_final.pth")
torch.save(discriminator_DCGAN.state_dict(), "custom_finetuned_discriminator_final.pth")
print("Fine-tuning Complete! Your GAN is now specialized to your dataset.")
```

## Diffusion Model

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import math
from torchvision.utils import make_grid
import matplotlib.pyplot as plt

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Hyperparameters
img_size = 128
channels = 3
condition_dim = 94
timesteps = 1000
batch_size = 128 # You can scale this up based on your GPU VRAM

# --- DDPM NOISE SCHEDULE ---
def linear_beta_schedule(timesteps):
    beta_start = 0.0001
    beta_end = 0.02
    return torch.linspace(beta_start, beta_end, timesteps)

betas = linear_beta_schedule(timesteps).to(device)
alphas = 1. - betas
alphas_cumprod = torch.cumprod(alphas, axis=0)

# Helper function to extract values for a specific timestep and reshape for broadcasting
def extract(a, t, x_shape):
    batch_size = t.shape[0]
    out = a.gather(-1, t)
    return out.reshape(batch_size, *((1,) * (len(x_shape) - 1)))
```

```python
# --- TIME & CONDITION EMBEDDING ---
class SinusoidalPositionEmbeddings(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.dim = dim

    def forward(self, time):
        half_dim = self.dim // 2
        embeddings = math.log(10000) / (half_dim - 1)
        embeddings = torch.exp(torch.arange(half_dim, device=device) * -embeddings)
        embeddings = time[:, None] * embeddings[None, :]
        embeddings = torch.cat((embeddings.sin(), embeddings.cos()), dim=-1)
        return embeddings

# --- U-NET BUILDING BLOCK ---

# class Attention(nn.Module):
#     def __init__(self, dim, heads=4, dim_head=32):
#         super().__init__()
#         self.scale = dim_head ** -0.5
#         self.heads = heads
#         hidden_dim = dim_head * heads
        
#         self.to_qkv = nn.Conv2d(dim, hidden_dim * 3, 1, bias=False)
#         self.to_out = nn.Conv2d(hidden_dim, dim, 1)

#     def forward(self, x):
#         b, c, h, w = x.shape
#         # Extract Queries, Keys, Values
#         qkv = self.to_qkv(x).chunk(3, dim=1)
#         q, k, v = map(lambda t: t.view(b, self.heads, -1, h * w).transpose(2, 3), qkv)

#         # Calculate Attention
#         dots = torch.matmul(q, k.transpose(-1, -2)) * self.scale
#         attn = dots.softmax(dim=-1)
        
#         # Apply Attention to Values
#         out = torch.matmul(attn, v)
#         out = out.transpose(2, 3).reshape(b, -1, h, w)
        
#         # Residual connection
#         return self.to_out(out) + x 

        
# class Block(nn.Module):
#     def __init__(self, in_ch, out_ch, time_emb_dim):
#         super().__init__()
#         self.conv1 = nn.Conv2d(in_ch, out_ch, kernel_size=3, padding=1)
#         self.conv2 = nn.Conv2d(out_ch, out_ch, kernel_size=3, padding=1)
#         self.bn1 = nn.GroupNorm(8, out_ch) # GroupNorm is standard for Diffusion, not BatchNorm
#         self.bn2 = nn.GroupNorm(8, out_ch)
#         self.time_mlp = nn.Linear(time_emb_dim, out_ch)
        
#     def forward(self, x, t_emb):
#         # First convolution
#         h = F.silu(self.bn1(self.conv1(x)))
#         # Inject Time & Condition embedding
#         time_emb = self.time_mlp(t_emb).unsqueeze(-1).unsqueeze(-1)
#         h = h + time_emb
#         # Second convolution
#         h = F.silu(self.bn2(self.conv2(h)))
#         return h

# # --- CONDITIONAL U-NET ---
# class ConditionalUNet(nn.Module):
#     def __init__(self):
#         super().__init__()
#         time_dim = 256
        
#         self.time_mlp = nn.Sequential(
#             SinusoidalPositionEmbeddings(time_dim),
#             nn.Linear(time_dim, time_dim),
#             nn.SiLU()
#         )
#         # Project your 94-dim condition to match the time embedding
#         self.cond_mlp = nn.Sequential(
#             nn.Linear(condition_dim, time_dim),
#             nn.SiLU()
#         )

#         # Downsample
#         self.down1 = Block(channels, 64, time_dim)
#         self.down2 = Block(64, 128, time_dim)
#         self.down3 = Block(128, 256, time_dim)
#         self.pool = nn.MaxPool2d(2)

#         # Bottleneck
#         self.bot1 = Block(256, 512, time_dim)
#         self.attn1 = Attention(512)
#         self.bot2 = Block(512, 256, time_dim)

#         # Upsample
#         self.up1 = nn.ConvTranspose2d(256, 128, kernel_size=2, stride=2)
#         self.up_block1 = Block(128 + 256, 128, time_dim) # Skip connection concatenation
        
#         self.up2 = nn.ConvTranspose2d(128, 64, kernel_size=2, stride=2)
#         self.up_block2 = Block(64 + 128, 64, time_dim)
        
#         self.up3 = nn.ConvTranspose2d(64, 64, kernel_size=2, stride=2)
#         self.up_block3 = Block(64 + 64, 64, time_dim)

#         self.out = nn.Conv2d(64, channels, kernel_size=1)

#     def forward(self, x, timestep, condition):
#         # Embed time and condition, then add them together
#         t_emb = self.time_mlp(timestep)
#         c_emb = self.cond_mlp(condition)
#         tc_emb = t_emb + c_emb 

#         # Down
#         x1 = self.down1(x, tc_emb)
#         p1 = self.pool(x1)
#         x2 = self.down2(p1, tc_emb)
#         p2 = self.pool(x2)
#         x3 = self.down3(p2, tc_emb)
#         p3 = self.pool(x3)

#         # Bottleneck
#         b = self.bot1(p3, tc_emb)
#         b = self.attn1(b)
#         b = self.bot2(b, tc_emb)

#         # Up with skip connections
#         u1 = self.up1(b)
#         u1 = torch.cat([u1, x3], dim=1)
#         u1 = self.up_block1(u1, tc_emb)

#         u2 = self.up2(u1)
#         u2 = torch.cat([u2, x2], dim=1)
#         u2 = self.up_block2(u2, tc_emb)
        
#         u3 = self.up3(u2)
#         u3 = torch.cat([u3, x1], dim=1)
#         u3 = self.up_block3(u3, tc_emb)

#         return self.out(u3)



"""NEW CODE"""
# --- IMPROVED RESNET DDPM BLOCKS ---
class ResnetBlock(nn.Module):
    def __init__(self, in_ch, out_ch, time_emb_dim):
        super().__init__()
        # Time & Condition MLP
        self.mlp = nn.Sequential(
            nn.SiLU(),
            nn.Linear(time_emb_dim, out_ch)
        )
        # First convolution block
        self.block1 = nn.Sequential(
            nn.GroupNorm(8, in_ch),
            nn.SiLU(),
            nn.Conv2d(in_ch, out_ch, 3, padding=1)
        )
        # Second convolution block
        self.block2 = nn.Sequential(
            nn.GroupNorm(8, out_ch),
            nn.SiLU(),
            nn.Conv2d(out_ch, out_ch, 3, padding=1)
        )
        # Residual connection path
        self.res_conv = nn.Conv2d(in_ch, out_ch, 1) if in_ch != out_ch else nn.Identity()

    def forward(self, x, t_emb):
        h = self.block1(x)
        # Inject time/condition embedding
        h = h + self.mlp(t_emb).unsqueeze(-1).unsqueeze(-1)
        h = self.block2(h)
        # True residual connection
        return h + self.res_conv(x)

class Attention(nn.Module):
    def __init__(self, dim, heads=4, dim_head=32):
        super().__init__()
        self.scale = dim_head ** -0.5
        self.heads = heads
        hidden_dim = dim_head * heads
        self.to_qkv = nn.Conv2d(dim, hidden_dim * 3, 1, bias=False)
        self.to_out = nn.Conv2d(hidden_dim, dim, 1)

    def forward(self, x):
        b, c, h, w = x.shape
        qkv = self.to_qkv(x).chunk(3, dim=1)
        q, k, v = map(lambda t: t.view(b, self.heads, -1, h * w).transpose(2, 3), qkv)
        dots = torch.matmul(q, k.transpose(-1, -2)) * self.scale
        attn = dots.softmax(dim=-1)
        out = torch.matmul(attn, v)
        out = out.transpose(2, 3).reshape(b, -1, h, w)
        return self.to_out(out) + x 

# --- GENERATIVE-GRADE CONDITIONAL U-NET ---
class ConditionalUNet(nn.Module):
    def __init__(self):
        super().__init__()
        time_dim = 512 
        
        self.time_mlp = nn.Sequential(
            SinusoidalPositionEmbeddings(time_dim),
            nn.Linear(time_dim, time_dim),
            nn.SiLU()
        )
        self.cond_mlp = nn.Sequential(
            nn.Linear(condition_dim, time_dim),
            nn.SiLU()
        )

        base_ch = 128 
        self.init_conv = nn.Conv2d(channels, base_ch, kernel_size=3, padding=1)

        # ==========================================
        #                 DOWNSAMPLE
        # ==========================================
        self.down1 = ResnetBlock(base_ch, base_ch, time_dim)
        
        self.down2_conv = nn.Conv2d(base_ch, base_ch * 2, 4, 2, 1) 
        self.down2 = ResnetBlock(base_ch * 2, base_ch * 2, time_dim)
        
        self.down3_conv = nn.Conv2d(base_ch * 2, base_ch * 4, 4, 2, 1)
        self.down3 = ResnetBlock(base_ch * 4, base_ch * 4, time_dim)

        # ==========================================
        #                 BOTTLENECK
        # ==========================================
        self.bot1 = ResnetBlock(base_ch * 4, base_ch * 4, time_dim)
        self.attn = Attention(base_ch * 4) 
        self.bot2 = ResnetBlock(base_ch * 4, base_ch * 4, time_dim)

        # ==========================================
        #                  UPSAMPLE
        # ==========================================
        self.up1_conv = nn.ConvTranspose2d(base_ch * 4, base_ch * 2, 4, 2, 1)
        # Expects: up1_conv out (256) + down2 out (256) = 512 channels
        self.up1 = ResnetBlock(base_ch * 4, base_ch * 2, time_dim) 
        
        self.up2_conv = nn.ConvTranspose2d(base_ch * 2, base_ch, 4, 2, 1)
        # Expects: up2_conv out (128) + down1 out (128) = 256 channels
        self.up2 = ResnetBlock(base_ch * 2, base_ch, time_dim)

        # FINAL OUTPUT
        self.out = nn.Conv2d(base_ch, channels, kernel_size=1)

    def forward(self, x, timestep, condition):
        t_emb = self.time_mlp(timestep) + self.cond_mlp(condition)

        x0 = self.init_conv(x)

        # Down
        d1 = self.down1(x0, t_emb)                  # Save for skip connection 2
        
        d2_down = self.down2_conv(d1)               
        d2 = self.down2(d2_down, t_emb)             # Save for skip connection 1
        
        d3_down = self.down3_conv(d2)               
        d3 = self.down3(d3_down, t_emb)             

        # Bottleneck
        b = self.bot1(d3, t_emb)                    
        b = self.attn(b)                            
        b = self.bot2(b, t_emb)                     

        # Up
        u1 = self.up1_conv(b)                       
        u1 = torch.cat([u1, d2], dim=1)             # Inject skip connection 1
        u1 = self.up1(u1, t_emb)                    

        u2 = self.up2_conv(u1)                      
        u2 = torch.cat([u2, d1], dim=1)             # Inject skip connection 2
        u2 = self.up2(u2, t_emb)                    

        return self.out(u2)
```

```python
# Forward process: Add noise to the real image
def q_sample(x_start, t, noise=None):
    if noise is None:
        noise = torch.randn_like(x_start)
    sqrt_alphas_cumprod_t = extract(torch.sqrt(alphas_cumprod), t, x_start.shape)
    sqrt_one_minus_alphas_cumprod_t = extract(torch.sqrt(1. - alphas_cumprod), t, x_start.shape)
    return sqrt_alphas_cumprod_t * x_start + sqrt_one_minus_alphas_cumprod_t * noise

# Reverse process: Generate an image from pure noise
@torch.no_grad()
def sample_images(model, num_images, condition):
    model.eval()
    # Start with pure Gaussian noise
    img = torch.randn((num_images, channels, img_size, img_size), device=device)
    
    # Iterate backward from T to 0
    for i in reversed(range(0, timesteps)):
        t = torch.full((num_images,), i, device=device, dtype=torch.long)
        predicted_noise = model(img, t, condition)
        
        alpha_t = extract(alphas, t, img.shape)
        alpha_bar_t = extract(alphas_cumprod, t, img.shape)
        beta_t = extract(betas, t, img.shape)
        
        # ====================================================
        # NEW FIX: x_0 Clipping to prevent value explosion
        # ====================================================
        # 1. Project the final noiseless image (x_0)
        pred_x0 = (img - torch.sqrt(1 - alpha_bar_t) * predicted_noise) / torch.sqrt(alpha_bar_t)
        
        # 2. Strictly clip it to the valid [-1, 1] range
        pred_x0 = torch.clamp(pred_x0, -1.0, 1.0)
        
        # 3. Recalculate the noise based on the safely clipped image
        predicted_noise = (img - torch.sqrt(alpha_bar_t) * pred_x0) / torch.sqrt(1 - alpha_bar_t)
        # ====================================================

        if i > 0:
            noise = torch.randn_like(img)
        else:
            noise = torch.zeros_like(img)
            
        # The DDPM reverse step math
        img = (1 / torch.sqrt(alpha_t)) * (img - ((1 - alpha_t) / torch.sqrt(1 - alpha_bar_t)) * predicted_noise) + torch.sqrt(beta_t) * noise
        
    model.train()
    
    # Un-normalize back to [0, 1] for viewing
    img = (img + 1) / 2
    return img.clamp(0, 1)
```

```python
import copy

# ==========================================
#         1. INITIALIZATION & EMA SETUP
# ==========================================
model = ConditionalUNet().to(device)

# Create the shadow model for EMA
ema_model = copy.deepcopy(model).eval()
ema_model.requires_grad_(False)
ema_decay = 0.995 # Standard decay rate for diffusion stabilization

# Diffusion usually trains better with a slightly higher LR than GANs
optimizer = optim.Adam(model.parameters(), lr=1e-4) 
loss_fn = nn.MSELoss()

epochs = 500

# ==========================================
#              2. TRAINING LOOP
# ==========================================
for epoch in range(epochs):
    epoch_loss = 0
    
    for i, (images, class_labels, country_labels, major_labels) in enumerate(dataloader):
        images = images.to(device)
        conditions = torch.cat([class_labels, country_labels, major_labels], dim=1).to(device)
        batch_size = images.shape[0]
        
        optimizer.zero_grad()
        
        # 1. Sample a random timestep for each image in the batch
        t = torch.randint(0, timesteps, (batch_size,), device=device).long()
        
        # 2. Generate random noise to add
        noise = torch.randn_like(images)
        
        # 3. Forward Process: Add noise to the real images based on timestep 't'
        x_noisy = q_sample(images, t, noise)
        
        # 4. Neural Network Pass: Predict the noise that was added
        predicted_noise = model(x_noisy, t, conditions)
        
        # 5. Calculate Loss
        loss = loss_fn(predicted_noise, noise)
        loss.backward()
        optimizer.step()
        
        # ----------------------------------------------------
        # 6. EMA UPDATE (Update the shadow model)
        # ----------------------------------------------------
        with torch.no_grad():
            for param_train, param_ema in zip(model.parameters(), ema_model.parameters()):
                param_ema.data.mul_(ema_decay).add_(param_train.data, alpha=1.0 - ema_decay)

        epoch_loss += loss.item()
        
    print(f"Epoch {epoch+1}/{epochs} | Average Loss: {epoch_loss / len(dataloader):.4f}")
    
    # ==========================================
    #              3. LOGGING & SAVING
    # ==========================================
    if (epoch + 1) % 50 == 0:
        # Create a batch of dummy conditions for the preview grid
        dummy_class = torch.zeros(16, 4, device=device)
        dummy_country = torch.zeros(16, 48, device=device)
        dummy_major = torch.zeros(16, 42, device=device)
        
        # Randomly populate one category per image for preview
        dummy_class.scatter_(1, torch.randint(0, 4, (16, 1), device=device), 1.0)
        dummy_country.scatter_(1, torch.randint(0, 48, (16, 1), device=device), 1.0)
        dummy_major.scatter_(1, torch.randint(0, 42, (16, 1), device=device), 1.0)
        
        preview_conds = torch.cat([dummy_class, dummy_country, dummy_major], dim=1)
        
        # CRITICAL: Pass the ema_model into your sample_images function, NOT the training model!
        sampled_imgs = sample_images(ema_model, 16, preview_conds) 
        
        grid = make_grid(sampled_imgs.cpu(), nrow=4, padding=2)
        plt.figure(figsize=(8, 8))
        plt.axis("off")
        plt.title(f"EMA Diffusion Generated Faces - Epoch {epoch+1}")
        plt.imshow(grid.permute(1, 2, 0))
        plt.show()

# Save BOTH models at the end of training
torch.save(model.state_dict(), "training_conditional_ddpm.pth")
torch.save(ema_model.state_dict(), "ema_ready_conditional_ddpm.pth") 

print("Training complete! Use 'ema_ready_conditional_ddpm.pth' for all future generation tasks.")
```
